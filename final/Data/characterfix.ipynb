## Quick fix for some characters in highcount_Use_hosted.csv

```python
import pandas as pd
```

```python
#import highcount_Use_hosted.csv
df = pd.read_csv('highcount_Use_hosted.csv')
df.head()
```

```python
#find all unique artist names
unique_artists = df['Artist'].unique()
print(unique_artists)
```

```python
#some artist names have weird characters, so we will fix those
df['Artist'] = df['Artist'].str.replace('\\u00e9', 'é')
df['Artist'] = df['Artist'].str.replace('\\u00ee', 'î')
df['Artist'] = df['Artist'].str.replace('\\u00eb', 'ë')
df['Artist'] = df['Artist'].str.replace('\\u00e7', 'ç')
df['Artist'] = df['Artist'].str.replace('Akin & Harrison\\nJames Akin', 'Akin & Harrison')





unique_artists = df['Artist'].unique()
print(unique_artists)
```

```python
# export dataframe to new csv
df.to_csv('highcount_Use_hosted.csv', index=False)
```
